```python
def AND(x1,x2):
  w1=1
  w2=1
  bias=-1.5
  activation=w1*x1+w2*x2+bias
  if activation>0:
    return 1
  else:
    return 0
res=AND(0,0)
r2=AND(1,1)
R3=AND(1,0)
r4=AND(0,1)
print(res)
print(r4)
print(R3)
print(r2)
```

```python
def OR(x1,x2):
  w1=1
  w2=1
  bias=-0.5
  activation=w1*x1+w2*x2+bias
  if activation>0:
    return 1
  else:
    return 0
res=OR(0,0)
r2=OR(1,1)
r3=OR(1,0)
r4=OR(0,1)
print(res)
print(r2)
print(r3)
print(r4)
```

```python
def NAND(x1,x2):
  w1=-1
  w2=-1
  bias=1.5
  activation=w1*x1+w2*x2+bias
  if activation>0:
    return 1
  else:
    return 0
r1=NAND(0,0)
r2=NAND(0,1)
r3=NAND(1,0)
r4=NAND(1,1)
print(r1)
print(r2)
print(r3)
print(r4)
```

```python
def XOR(x1,x2):
  h1=NAND(x1,x2)
  h2=OR(x1,x2)
  y=AND(h1,h2)
  return y
r1=XOR(0,0)
r2=XOR(0,1)
r3=XOR(1,0)
r4=XOR(1,1)
print(r1)
print(r2)
print(r3)
print(r4)
```

```python
def NOR(x1,x2):
  w1=-1
  w2=-1
  bias=0.5
  activation=w1*x1+w2*x2+bias
  if activation>0:
    return 1
  else:
    return 0
r1=NOR(0,0)
r2=NOR(0,1)
r3=NOR(1,0)
r4=NOR(1,1)
print(r1)
print(r2)
print(r3)
print(r4)
```

```python
def XNOR(x1,x2):
  h1=NOR(x1,x2)
  h2=AND(x1,x2)
  y=OR(h1,h2)
  return y
r1=XNOR(0,0)
r2=XNOR(0,1)
r3=XNOR(1,0)
r4=XNOR(1,1)
print(r1)
print(r2)
print(r3)
print(r4)
```

```python
#implement AND using perceptron
import numpy as np
def step(x):
  return 1 if x>=0 else 0
class perceptron:
  def __init__(self,weight,bias):
    self.weight=weight
    self.bias=bias
  def predict(self,inputs):
    total=np.dot(self.weight,inputs)+self.bias
    print("THRESHOLD VALUE:",total)
    return step(total)
weights=np.array([1,1])
bias=-1.5
p=perceptron(weights,bias)
print("AND")
print(p.predict([0,0]))
print(p.predict([0,1]))
print(p.predict([1,0]))
print(p.predict([1,1]))
```

```python
#multi perceptron
def step(x):
  return 1 if x>=0 else 0
class mlp:
  def __init__(self):
    self.w1=np.array([1,1])
    self.b1=-1.5
    self.w2=np.array([1,1])
    self.b2=-0.5
    self.w3=np.array([-1,-1])
    self.b3=1.5
  def predict(self,inputs):
    self.h1=step(np.dot(self.w3,inputs)+self.b3)
    self.h2=step(np.dot(self.w2,inputs)+self.b2)
    self.o1=step(np.dot([self.h1,self.h2],self.w1)+self.b1)
    return self.o1
xor=mlp()
for x in [(0,0),(0,1),(1,0),(1,1)]:
  print(x,xor.predict(x))
```

```python
import numpy as np
import pandas as pd

def step(x):
    return 1 if x >= 0 else 0

class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=20):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = 0

    def predict(self, x):
        return step(np.dot(x, self.weights) + self.bias)

    def train(self, X, y):
        self.weights = np.zeros(X.shape[1])

        for epoch in range(self.epochs):
            errors = 0
            print(f"\nEpoch {epoch + 1}")

            for i in range(len(X)):
                y_pred = self.predict(X[i])
                error = y[i]- y_pred

                if error != 0:
                    self.weights += self.lr * error * X[i]
                    self.bias += self.lr * error
                    errors += 1

                    print(f"Updated weights: {self.weights}, Bias: {self.bias}")

            print(f"Epoch {epoch + 1} errors: {errors}")
            if errors == 0:
                print("Convergence")
                break
```
