```python
def AND(x1,x2):
  w1=1
  w2=1
  bias=-1.5
  activation=w1*x1+w2*x2+bias
  if activation>0:
    return 1
  else:
    return 0
res=AND(0,0)
r2=AND(1,1)
R3=AND(1,0)
r4=AND(0,1)
print(res)
print(r4)
print(R3)
print(r2)
```

```python
def OR(x1,x2):
  w1=1
  w2=1
  bias=-0.5
  activation=w1*x1+w2*x2+bias
  if activation>0:
    return 1
  else:
    return 0
res=OR(0,0)
r2=OR(1,1)
r3=OR(1,0)
r4=OR(0,1)
print(res)
print(r2)
print(r3)
print(r4)
```

```python
def NAND(x1,x2):
  w1=-1
  w2=-1
  bias=1.5
  activation=w1*x1+w2*x2+bias
  if activation>0:
    return 1
  else:
    return 0
r1=NAND(0,0)
r2=NAND(0,1)
r3=NAND(1,0)
r4=NAND(1,1)
print(r1)
print(r2)
print(r3)
print(r4)
```

```python
def XOR(x1,x2):
  h1=NAND(x1,x2)
  h2=OR(x1,x2)
  y=AND(h1,h2)
  return y
r1=XOR(0,0)
r2=XOR(0,1)
r3=XOR(1,0)
r4=XOR(1,1)
print(r1)
print(r2)
print(r3)
print(r4)
```

```python
def NOR(x1,x2):
  w1=-1
  w2=-1
  bias=0.5
  activation=w1*x1+w2*x2+bias
  if activation>0:
    return 1
  else:
    return 0
r1=NOR(0,0)
r2=NOR(0,1)
r3=NOR(1,0)
r4=NOR(1,1)
print(r1)
print(r2)
print(r3)
print(r4)
```

```python
def XNOR(x1,x2):
  h1=NOR(x1,x2)
  h2=AND(x1,x2)
  y=OR(h1,h2)
  return y
r1=XNOR(0,0)
r2=XNOR(0,1)
r3=XNOR(1,0)
r4=XNOR(1,1)
print(r1)
print(r2)
print(r3)
print(r4)
```

```python
#implement AND using perceptron
import numpy as np
def step(x):
  return 1 if x>=0 else 0
class perceptron:
  def __init__(self,weight,bias):
    self.weight=weight
    self.bias=bias
  def predict(self,inputs):
    total=np.dot(self.weight,inputs)+self.bias
    print("THRESHOLD VALUE:",total)
    return step(total)
weights=np.array([1,1])
bias=-1.5
p=perceptron(weights,bias)
print("AND")
print(p.predict([0,0]))
print(p.predict([0,1]))
print(p.predict([1,0]))
print(p.predict([1,1]))
```

```python
#multi perceptron
def step(x):
  return 1 if x>=0 else 0
class mlp:
  def __init__(self):
    self.w1=np.array([1,1])
    self.b1=-1.5
    self.w2=np.array([1,1])
    self.b2=-0.5
    self.w3=np.array([-1,-1])
    self.b3=1.5
  def predict(self,inputs):
    self.h1=step(np.dot(self.w3,inputs)+self.b3)
    self.h2=step(np.dot(self.w2,inputs)+self.b2)
    self.o1=step(np.dot([self.h1,self.h2],self.w1)+self.b1)
    return self.o1
xor=mlp()
for x in [(0,0),(0,1),(1,0),(1,1)]:
  print(x,xor.predict(x))
```

```python
import numpy as np
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
functions = {
    "f1":  [0,0,0,0],
    "f2":  [0,0,0,1], #AND
    "f3":  [0,0,1,0],
    "f4":  [0,0,1,1],
    "f5":  [0,1,0,0],
    "f6":  [0,1,0,1],
    "f7":  [0,1,1,0], #XOR
    "f8":  [0,1,1,1],
    "f9":  [1,0,0,0],
    "f10": [1,0,0,1],#XNOR
    "f11": [1,0,1,0],
    "f12": [1,0,1,1],
    "f13": [1,1,0,0],
    "f14": [1,1,0,1],
    "f15": [1,1,1,0],
    "f16": [1,1,1,1]
}
```

```python
def train_perceptron(X, y, lr=0.1, epochs=50):
    weights = np.zeros(X.shape[1])
    bias = 0

    for _ in range(epochs):
        errors = 0
        for i in range(len(X)):
            net = np.dot(X[i], weights) + bias
            pred = 1 if net >= 0 else 0
            error = y[i] - pred
            if error != 0:
                weights += lr * error * X[i]
                bias += lr * error
                errors += 1
        if errors == 0:
            return True  # Converged
    return False  # Did not converge
```

```python
linearly_separable = []
not_separable = []

for name, y_vals in functions.items():
    y = np.array(y_vals)
    if train_perceptron(X, y):
        linearly_separable.append(name)
    else:
        not_separable.append(name)

print("Linearly separable functions:")
print(linearly_separable)

print("\nNot linearly separable functions:")
print(not_separable)
```

```python
def step(x):
    return 1 if x >= 0 else 0
import numpy as np


def hidden_layer(x1, x2):
    h1 = step(-x1 - x2 + 0.5)
    h2 = step(-x1 + x2 - 0.5)
    h3 = step( x1 - x2 - 0.5)
    h4 = step( x1 + x2 - 1.5)
    return [h1, h2, h3, h4]
```

```python
def mlp_boolean(x1, x2, f):
    h = hidden_layer(x1, x2)
    weights = np.array(f)
    bias = -0.5
    net = np.dot(weights, h) + bias
    return step(net)
```

```python
AND = [0, 0, 0, 1]
inputs = [(0,0), (0,1), (1,0), (1,1)]

print("AND Function:")
for x in inputs:
    print(x, "→", mlp_boolean(x[0], x[1], AND))
```

```python
XOR = [0, 1, 1, 0]

print("\nXOR Function:")
for x in inputs:
    print(x, "→", mlp_boolean(x[0], x[1], XOR))
```
